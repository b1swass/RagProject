```python
from pydantic import BaseModel,Field
from typing import List,Annotated
```

```python
class User(BaseModel):
    id: int
    name: str
    email: str
    is_active: bool = True  

```

```python
user=User(id=1, name="John Doe", email="john.doe@example.com")
user
```

```python
#field validation
class Product(BaseModel):
    title: str=Field(..., min_length=3, max_length=50) 
    price: float=Field(..., gt=0)
    description: str = Field(..., min_length=10, max_length=50)
```

```python
product=Product(
    title="Sample Product",
    price=19.99,
    description="This is a sample product description.")
product2=Product(
    title="Sample Product 2",
    price=29.99,
    description="This is a sample product description 2.")
```

multiple class validation

```python
class Order(BaseModel):
    order_id: str=Field(..., min_length=5, max_length=20)
    user: User=Field(...)
    product: Product=Field(...)
    quantity: int = Field(..., gt=0)
    total_price: float = Field(..., gt=0)

order=Order(
    order_id="ORD12345",
    user=user,
    product=product,
    quantity=2,
    total_price=product.price * 2
)
print(order)
```

```python
print(order.product)
print("\n\n")
print(order.user)
```

```python
class Carts(BaseModel):
    cart_id: str=Field(..., min_length=5, max_length=20)
    items: list[Product]
    comment:str=Field(default="")
```

```python
cart= Carts(
    cart_id="CART12345",
    items=[product, product2],
    comment="This is a sample cart."
)
print(cart)
```

```python
for product in cart.items:
    print(product.title, product.price)
```
